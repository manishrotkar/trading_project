```python
import sqlite3
import pandas as pd

db = "../database/futures.db"

conn = sqlite3.connect(db)

df = pd.read_sql_query(
    """
    SELECT *
    FROM futures_data
    WHERE symbol = 'NSE:NIFTY26SEPFUT'
    ORDER BY datetime
    """,
    conn
)

conn.close()

df.head()
```

```python
duplicates = df[
    df.duplicated(
        subset=["symbol", "datetime"],
        keep=False
    )
]

print("Duplicate candle rows:", len(duplicates))
print(
    "Duplicate timestamps:",
    duplicates[["symbol", "datetime"]]
    .drop_duplicates()
    .shape[0]
)
```

```python
check = pd.read_sql_query(
    """
    SELECT contract, COUNT(*) AS rows
    FROM futures_data
    WHERE symbol = 'NSE:NIFTY26SEPFUT'
    GROUP BY contract
    """,
    sqlite3.connect("../database/futures.db")
)

check
```

```python
import sqlite3

db = "../database/futures.db"

conn = sqlite3.connect(db)

conn.execute(
    """
    DELETE FROM futures_data
    WHERE symbol = 'NSE:NIFTY26SEPFUT'
      AND contract = 'NIFTY 29 Sep 26 FUT'
    """
)

conn.commit()

print("Duplicate NIFTY September rows deleted.")
print("Rows deleted:", conn.total_changes)

conn.close()
```
